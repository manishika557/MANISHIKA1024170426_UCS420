Find the most frequent value and their indice(s) in the following arrays:
i. x = np.array([1,2,3,4,5,1,2,1,1,1])
ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]

print("Most frequent value:", most_frequent)
print("Indices:", np.where(x == most_frequent)[0])
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]

print("Most frequent value:", most_frequent)
print("Indices:", np.where(y == most_frequent)[0])
```

4. Write program to create an 1-D NumPy array named <<Your Name>> with evenly
spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array,
shape, total elements, the data type of each element and total number of bytes consumed
by the array. Find the transpose of this array using reshape() attribute. Can we do the same
with T attribute?

```python
# creating the array
Manishika = np.linspace(10, 100, 25)
print(Manishika)
```

```python
# dimension
print("Dimensions:", Manishika.ndim)
```

```python
# shape
print("Shape:", Manishika.shape)
```

```python
# total element
print("Total elements:", Manishika.size)
```

```python
# data type
print("Data type:", Manishika.dtype)
```

```python
# total bytes
print("Total bytes:", Manishika.nbytes)
```

```python
# transpose
transpose = Manishika.reshape(25, 1)
print(transpose)
```

```python
# transpose using t
print(Manishika.T)
```

5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>>
with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean,
median, max, min, unique elements. Reshape the array to four rows and three columns and
name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three
columns and name it as resized_ ucs420_<your_name>>.

```python
ucs420_Manishika = np.array([[10, 20, 30, 40],
                             [50, 60, 70, 80],
                             [90, 15, 20, 35]])

print(ucs420_Manishika)
```

```python
# mean
print("Mean:", np.mean(ucs420_Manishika))
```

```python
# median
print("Median:", np.median(ucs420_Manishika))
```

```python
# maximum
print("Maximum:", np.max(ucs420_Manishika))
```

```python
# minimum
print("Minimum:", np.min(ucs420_Manishika))
```

```python
# unique elements
print("Unique elements:", np.unique(ucs420_Manishika))
```

```python
# reshape
reshaped_ucs420_Manishika = ucs420_Manishika.reshape(4, 3)
print(reshaped_ucs420_Manishika)
```

```python
# resize
resized_ucs420_Manishika = np.resize(ucs420_Manishika, (2, 3))
print(resized_ucs420_Manishika)
```
