Calculate the following statistics from the dataset:
1. The average salary of employees.
2. The total bonus paid to all employees.
3. The youngest employee's age.
4. The highest performance rating.

```python
print("\ni) Average Salary:")
print(df["Salary"].mean())

print("\nii) Total Bonus:")
print(df["Bonus"].sum())

print("\niii) Youngest Age:")
print(df["Age"].min())

print("\niv) Highest Rating:")
print(df["Rating"].max())
```

Sort the DataFrame by the Salary column in descending order.

```python
print(df.sort_values(by="Salary", ascending=False))
```

Add a new column that categorizes employees based on their performance rating:
i. Excellent for ratings >= 4.5
ii. Good for ratings >= 4.0 but < 4.5
iii. Average for ratings < 4.0

```python
def performance(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

df["Performance"] = df["Rating"].apply(performance)

print("\n Performance Category:")
print(df)
```

Identify missing values in the DataFrame.

```python
print(df.isnull().sum())
```

Rename the Employee_ID column to ID.

```python
df.rename(columns={"Employee_ID": "ID"}, inplace=True)

print("\n Renamed Column:")
print(df)
```

Find all employees who:
i. Have more than 5 years of experience.
ii. Belong to the IT department.

```python
print("\n More than 5 years of experience:")
print(df[df["Years_of_Experience"] > 5])

print("\n IT Department employees:")
print(df[df["Department"] == "IT"])
```

Modify the dataset by adding a new column, Tax, which deducts 10% of the
Salary.

```python
df["Tax"] = df["Salary"] * 0.10

print(df)
```

Save the modified DataFrame (with added columns) to a new CSV file.

```python
df.to_csv("employees_modified.csv", index=False)
```
