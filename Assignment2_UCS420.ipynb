Set the random seed to your own roll number using random.seed(your_roll_number) before generating
your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed
produces a different list, so this cannot be copied from a classmate).
i. ii. iii. iv. v. Generate a list of 100 random numbers between 100 and 900 (inclusive).
Count and print all odd numbers in the list.
Count and print all even numbers in the list.
Count and print all prime numbers in the list, and additionally build the actual list of prime numbers
found using list comprehension.
Print the number that occurs most frequently in your list, and how many times it occurs.

```python
import random

random.seed(1024170426)

numbers = [random.randint(100, 900) for _ in range(100)]

odd_numbers = [num for num in numbers if num % 2 != 0]
even_numbers = [num for num in numbers if num % 2 == 0]

def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True

prime_numbers = [num for num in numbers if is_prime(num)]

most_frequent = max(numbers, key=numbers.count)
frequency = numbers.count(most_frequent)

print("Random numbers:", numbers)
print("Odd numbers:", odd_numbers)
print("Count of odd numbers:", len(odd_numbers))
print("Even numbers:", even_numbers)
print("Count of even numbers:", len(even_numbers))
print("Prime numbers:", prime_numbers)
print("Count of prime numbers:", len(prime_numbers))
print("Most frequent number:", most_frequent)
print("Frequency:", frequency)
```

Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for
each digit} and set B = {digit × 9 for each digit}. Print both sets.
vi. vii. Find the union of A and B — all unique values across both sets.
Find the intersection of A and B — values common to both.
viii. Find A − B and B − A separately (not combined) using difference(), and explain in one line how this
differs from symmetric_difference().
ix. x. xi. Find the symmetric difference of A and B.
Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset().
Ask the user to input a value X. Remove X from set A if present using discard() (not remove()) — explain
in one line why discard() is safer than remove() when you're not sure the value exists.

```python
roll_number = input("Enter your 8-digit roll number: ")

digits = [int(digit) for digit in roll_number]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)

union_set = A.union(B)
print("Union:", union_set)

intersection_set = A.intersection(B)
print("Intersection:", intersection_set)

A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

symmetric_diff = A.symmetric_difference(B)
print("Symmetric difference:", symmetric_diff)

print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

X = int(input("Enter a value to remove from A: "))

A.discard(X)

print("Set A after discard:", A)
```
