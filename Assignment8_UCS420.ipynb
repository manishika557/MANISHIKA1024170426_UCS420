**QUESTION 1**

Import libraries

```python

import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
```

```python

text = """Ice cream is my favorite dessert because it is sweet and delicious.
I love eating chocolate and vanilla ice cream with my friends.
During summer, a cold scoop of ice cream makes me feel happy.
There are many flavors, such as strawberry, mango, and butterscotch.
I enjoy trying new flavors whenever I visit an ice cream shop.
Ice cream makes ordinary days more enjoyable and special."""

print(text)
```

Convert to lowercase and remove punctuation

```python

clean_text = re.sub(r'[^a-zA-Z\s]', '', text.lower())
print(clean_text)
```

Tokenize into words and sentences

```python

words = word_tokenize(text.lower())
sentences = sent_tokenize(text)

print("Words:", words)
print("Sentences:", sentences)
```

Compare split() and word_tokenize()

```python

print("Using split():", text.lower().split())
print("Using word_tokenize():", word_tokenize(text.lower()))
```

Remove stopwords

```python

stop_words = set(stopwords.words('english'))
filtered_words = [word for word in word_tokenize(clean_text)
                  if word not in stop_words]

print(filtered_words)
```

Word frequency distribution

```python

freq = FreqDist(filtered_words)
print(freq.most_common())
```

**QUESTION 2**

Import libraries

```python

from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)
```

Extract alphabet-only words and remove stopwords

```python

alphabet_words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
filtered_words = [w for w in alphabet_words if w not in stop_words]

print("Alphabet-only words:", alphabet_words)
print("After stopword removal:", filtered_words)
```

Perform stemming

```python

ps = PorterStemmer()
stemmed_words = [ps.stem(w) for w in filtered_words]

print(stemmed_words)
```

Perform lemmatization

```python

lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(w) for w in filtered_words]

print(lemmatized_words)
```

Compare the outputs

```python

print("Original words: ", filtered_words)
print("Stemmed words:  ", stemmed_words)
print("Lemmatized words:", lemmatized_words)
```

Observation:
- Stemming removes word endings to produce a basic stem, which may not be a proper English word.
- Lemmatization converts words to their meaningful dictionary forms.
- Stemming is useful when speed is important, while lemmatization is preferable when meaningful words are require

**QUESTION 3**

Import libraries

```python

import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

Create three short texts

```python

texts = [
    "Chocolate ice cream is creamy sweet and delicious.",
    "Vanilla ice cream has a smooth creamy taste.",
    "Mango ice cream is refreshing sweet and fruity."
]

for i, t in enumerate(texts, 1):
    print(f"Text {i}: {t}")
```

Generate Bag of Words

```python

count_vectorizer = CountVectorizer(stop_words='english')
bow = count_vectorizer.fit_transform(texts)

bow_df = pd.DataFrame(
    bow.toarray(),
    columns=count_vectorizer.get_feature_names_out(),
    index=["Text 1", "Text 2", "Text 3"]
)

print(bow_df)
```

Calculate TF-IDF scores

```python

tfidf_vectorizer = TfidfVectorizer(stop_words='english')
tfidf = tfidf_vectorizer.fit_transform(texts)

tfidf_df = pd.DataFrame(
    tfidf.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out(),
    index=["Text 1", "Text 2", "Text 3"]
)

print(tfidf_df.round(3))
```

Print top 3 keywords from each text

```python

for i, row in enumerate(tfidf.toarray(), 1):
    top_indices = row.argsort()[-3:][::-1]
    keywords = [
        (tfidf_vectorizer.get_feature_names_out()[j], round(row[j], 3))
        for j in top_indices
    ]
    print(f"Text {i} top keywords:", keywords)
```

Observation: Bag of Words counts how often each word appears. TF-IDF gives more importance to words that help distinguish one text from the others. The top keywords should highlight the main flavor or qualities discussed in each review.

**QUESTION 4**

Write two technology descriptions

```python

text1 = """Artificial intelligence helps computers learn from data.
AI can recognize images and understand human language.
It is used in healthcare, education, and automation.
AI systems can make predictions and solve problems."""

text2 = """Blockchain stores information in connected digital blocks.
It provides secure and transparent data records.
Blockchain is used in finance and digital transactions.
It can improve trust and protect stored information."""

print("Text 1:\n", text1)
print("\nText 2:\n", text2)
```

Preprocess and tokenize

```python

import re
from nltk.tokenize import word_tokenize

clean1 = re.sub(r'[^a-zA-Z\s]', '', text1.lower())
clean2 = re.sub(r'[^a-zA-Z\s]', '', text2.lower())

tokens1 = word_tokenize(clean1)
tokens2 = word_tokenize(clean2)

print("Technology 1 tokens:", tokens1)
print("Technology 2 tokens:", tokens2)
```

Calculate Jaccard similarity

```python

set1 = set(tokens1)
set2 = set(tokens2)

jaccard = len(set1 & set2) / len(set1 | set2)

print("Jaccard Similarity:", round(jaccard, 3))
```

Calculate cosine similarity using TF-IDF

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer(stop_words='english')
vectors = vectorizer.fit_transform([text1, text2])

cosine = cosine_similarity(vectors[0:1], vectors[1:2])[0][0]

print("Cosine Similarity:", round(cosine, 3))
```

Compare the metrics

```python

print("Jaccard Similarity:", round(jaccard, 3))
print("Cosine Similarity:", round(cosine, 3))

if cosine > jaccard:
    print("Cosine similarity gives better insight because it considers TF-IDF word weights.")
else:
    print("Jaccard similarity gives a direct comparison of shared and total unique words.")
```

Observation: Jaccard compares the overlap between unique words. Cosine similarity compares TF-IDF-weighted text vectors, so it can provide a more informative comparison when word importance matters. Since AI and blockchain discuss different technologies, both scores are expected to be relatively low.

Create product reviews

```python

from textblob import TextBlob
import matplotlib.pyplot as plt
from wordcloud import WordCloud

reviews = [
    "The ice cream is delicious and wonderfully creamy.",
    "I love the rich chocolate flavor and excellent taste.",
    "The ice cream was terrible and tasted very bad.",
    "The service was slow and the ice cream was disappointing.",
    "The ice cream was okay and the taste was average."
]

for review in reviews:
    print(review)
```

Find polarity and subjectivity

```python

for review in reviews:
    analysis = TextBlob(review)
    print("\nReview:", review)
    print("Polarity:", round(analysis.sentiment.polarity, 3))
    print("Subjectivity:", round(analysis.sentiment.subjectivity, 3))
```

Classify reviews

```python

for review in reviews:
    polarity = TextBlob(review).sentiment.polarity

    if polarity > 0:
        sentiment = "Positive"
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print(sentiment, ":", review)
```

Create a word cloud from positive reviews

```python

positive_reviews = [
    review for review in reviews
    if TextBlob(review).sentiment.polarity > 0
]

positive_text = " ".join(positive_reviews)

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white"
).generate(positive_text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```

Observation:
- Polarity measures sentiment, from −1 (negative) to +1 (positive).
- Subjectivity measures whether the text is opinion-based, from 0 (objective) to 1 (subjective).
- Reviews are classified according to their polarity.
- The word cloud displays frequently occurring words from positive reviews, with larger words appearing more prominently.

**QUESTION 6**

Import libraries

```python

import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

Create training text (~100 words)

```python

training_text = """
Ice cream is a popular dessert enjoyed by people around the world.
It comes in many flavors such as chocolate vanilla strawberry and mango.
People enjoy ice cream during summer because it feels cool and refreshing.
Some people prefer chocolate ice cream while others love vanilla ice cream.
Ice cream can be served in a cone or in a bowl.
Many shops prepare fresh ice cream using milk cream sugar and fruits.
Families often enjoy ice cream together after dinner.
A scoop of ice cream can make a normal day feel special.
Homemade ice cream allows people to try different flavors.
Ice cream is a sweet treat that brings happiness to many people.
"""
print(training_text)
```

Tokenize the text

```python

tokenizer = Tokenizer()
tokenizer.fit_on_texts([training_text])

total_words = len(tokenizer.word_index) + 1
sequences = []

for line in training_text.strip().split('\n'):
    token_list = tokenizer.texts_to_sequences([line])[0]
    for i in range(2, len(token_list) + 1):
        sequences.append(token_list[:i])

max_length = max(len(seq) for seq in sequences)
padded_sequences = pad_sequences(sequences, maxlen=max_length, padding='pre')

print("Vocabulary size:", total_words)
print("Total sequences:", len(sequences))
print("Padded sequence shape:", padded_sequences.shape)
```

Create input and output sequences

```python

X = padded_sequences[:, :-1]
y = padded_sequences[:, -1]

print("Input shape:", X.shape)
print("Output shape:", y.shape)
```

Build the LSTM model

```python

model = Sequential([
    Embedding(total_words, 10, input_length=max_length - 1),
    LSTM(50),
    Dense(total_words, activation='softmax')
])

model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.summary()
```

Train the model

```python

model.fit(X, y, epochs=100, verbose=0)
print("Model training completed.")
```

Generate 3 lines of new text

```python

def generate_text(seed_text, next_words=10):
    result = seed_text

    for _ in range(next_words):
        sequence = tokenizer.texts_to_sequences([result])[0]
        sequence = pad_sequences([sequence], maxlen=max_length - 1, padding='pre')
        prediction = model.predict(sequence, verbose=0)
        next_index = np.argmax(prediction)

        if next_index == 0:
            break

        next_word = tokenizer.index_word.get(next_index, "")
        result += " " + next_word

    return result

print(generate_text("ice cream", 10))
print(generate_text("chocolate ice cream", 10))
print(generate_text("people enjoy", 10))
```

Observation: The tokenizer converts words into numbers, and the LSTM learns patterns from the training sequences to predict the next word. The model then generates text from the given seed phrases. Because the training dataset is small, the generated sentences may repeat words or may not be grammatically correct.
