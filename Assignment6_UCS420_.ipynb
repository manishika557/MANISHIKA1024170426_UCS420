 1. Consider the following sensor readings:
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
A sensor has an average measurement error of +2°C.
Perform the following without using a for loop:
Add 2°C to every reading using vectorization

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print(corrected_temperature)
```

Convert all temperatures from Celsius to Fahrenheit

```python
fahrenheit = (temperature * 9 / 5) + 32
print(fahrenheit)
```

Identify all readings greater than 32°C using Boolean indexing.

```python
readings = temperature[temperature > 32]
print(readings)
```

Count how many readings exceed 32°C.

```python
count = np.sum(temperature > 32)
print(count)
```

 Explain how vectorization and Boolean indexing can make data processing more efficient
than explicitly iterating through every value.

Vectorization applies an operation to the entire array at once without using a for loop.
Boolean indexing selects elements that satisfy a condition directly.
Both methods are faster and more efficient than explicitly iterating through every value.

2. The following matrix represents the number of daily steps recorded by four users
over three days:

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print(steps)
```

Calculate the total steps recorded.

```python
print("Total steps:", np.sum(steps))
```

Calculate the mean number of steps.

```python
print("Mean steps:", np.mean(steps))
```

Find the maximum and minimum values.

```python
print("Maximum:", np.max(steps))
print("Minimum:", np.min(steps))
```

Calculate the total steps for each day using axis=0.

```python
print("Total steps for each day:", np.sum(steps, axis=0))
```

Calculate the total steps for each user using axis=1.

```python
print("Total steps for each user:", np.sum(steps, axis=1))
```

Find the user/day position containing the maximum number of steps using argmax()

```python
position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum:", position)
```

4. A simple cognitive-assistive system wants to predict a user's assistance score based
on three factors:
• Sleep hours
• Activity level
• Stress level

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("X:")
print(X)

print("y:")
print(y)
```

Display the shape and dimensions of X.

```python
print("Shape:", X.shape)
print("Dimensions:", X.ndim)
```

Calculate: X.T and explain what the transpose represents.

```python
print(X.T)
```

Calculate the matrix product: X.T @ X

```python
product = X.T @ X
print(product)
```

Calculate the matrix inverse using: np.linalg.inv() where appropriate.

```python
inverse = np.linalg.inv(X.T @ X)
print(inverse)
```

Implement the Ordinary Least Squares equation: 𝛽 = (𝑋!𝑋)"#𝑋!𝑦 using NumPy.

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("Coefficients:", beta)
```

Explain what the resulting coefficients represent in the context of the three user
features.

The first coefficient represents the effect of sleep hours on the assistance score.
The second coefficient represents the effect of activity level on the assistance score.
The third coefficient represents the effect of stress level on the assistance score.

A new user has: new_user = np.array([5, 40, 7]) Use your calculated coefficients to
obtain a predicted assistance score.

```python
new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("Predicted assistance score:", predicted_score)
```
