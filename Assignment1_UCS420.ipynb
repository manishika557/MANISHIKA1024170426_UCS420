```python
# WAP to add all numbers divisible by 7 and 9 from 1 to n and n is given by the user.
n = int(input("Enter a number: "))

total = 0

for i in range(1, n + 1):
    if i % 7 == 0 or i % 9 == 0:
        total += i

print("Sum =", total)
```

```python
# WAP to add all prime numbers from 1 to n and n is given by the user.
n = int(input("Enter a number: "))

total = 0

for i in range(2, n + 1):
    prime = True
    for j in range(2, i):
        if i % j == 0:
            prime = False
            break
    if prime:
        total += i

print("Sum =", total)
```

```python
#  WAP using function that add all odd numbers from 1 to n, n is given by the user.
def add_odd(n):
    total = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            total += i
    return total

n = int(input("Enter a number: "))
print("Sum =", add_odd(n))
```

```python
# WAP using function that add all prime numbers from 1 to n, n given by the user.
def add_prime(n):
    total = 0

    for i in range(2, n + 1):
        prime = True
        for j in range(2, i):
            if i % j == 0:
                prime = False
                break
        if prime:
            total += i

    return total

n = int(input("Enter a number: "))
print("Sum =", add_prime(n))
```
